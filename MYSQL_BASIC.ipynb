# Q9. What is the Difference Between a Primary Key and a Foreign Key?

ANSWER

1. FEATURE OF PRIMARY KEY
* PURPOSE =  Uniquely identifies each record in a table
* UNIQUENESS = MUST BE UNIQUE
* NULL  ALLOWED = NOT ALLLOWED

2. FEATURE OF FOREIGN KEY

* PURPOSE = Links records between two tables
* UNIQUENESS = Can have duplicates
* NULL ALLOWED = Allowed depending on design


# Q10. What Are Constraints in SQL and Why Are They Used?


ANSWER

- PRIMARY KEY: Ensures uniqueness and non-null values.
- FOREIGN KEY: Maintains referential integrity between tables.
- NOT NULL: Prevents null values.
- UNIQUE: Ensures all values in a column are different.
- CHECK: Validates data based on a condition.
- DEFAULT: Sets a default value if none is provided.

2. Why they're used:
To prevent invalid data entry, enforce business rules, and maintain consistency across the database.
