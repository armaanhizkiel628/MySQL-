# Q1. What is a Common Table Expression (CTE), and how does it improve SQL query readability?

ANSWER

1. A Common Table Expression  is a temporary, named result set in SQL that you define within a query using the with  keyword. It exists only for  duration of that query and can be referenced like a table or view.

2. how does it improve ?


*   Breaks complex queries into logical parts  Instead of nesting multiple subqueries a CTE allows you to separate steps clearly.

*   Makes queries easier to understand and maintain Each CTE has a meaningful name, which explains what the query step is doing without reading all the logic.









# Q2. Why are some views updatable while others are read-only? Explain with an example.

answer

1. SQL views are updatable while others are read-only because of how much the database can clearly  map a change in the view back to a single row in a base table. If the database cannot clearly determine which table and which row should be updated, the view becomes read-only.

2. When a View Is Updatable
* Does not include calculated or derived columns
* Is based on a single table

* example A view is updatable only when each row in the view clearly corresponds to one row in the original table.


3. When a View Is Read-Only
* A view becomes read-only when it includes elements that make row mapping ambiguous, such as:
* Joins across multiple tables
* Aggregate functions
* GROUP BY

4. example
* Some views are read-only because the database cannot clearly identify which row in the base table should be updated.



#3  What advantages do stored procedures offer compared to writing raw SQL queries repeatedly?

answer
1. Stored procedures have many advantages compared to writing SQL queries again and again.
* Can be reused a stored procedure is written once and can be used many times, so we do not need to write the same SQL query repeatedly

* Faster execution stored procedures are already saved and processed by the database, so they run faster than writing new SQL queries each time.

* Easy to maintain if we need to change the logic, we only update the stored procedure instead of changing SQL queries in many places.

* More secure users can run stored procedures without directly accessing database tables, which keeps data safe.

* Stored procedures save time, improve speed, increase security, and make database work easier.

# 4 What is the purpose of triggers in a database? Mention one use case where a trigger is essential

answer

1. Triggers are used in a database to automatically perform an action when a specific event happens on a table, such as INSERT, UPDATE, or DELETE.

2. The main purpose of a trigger is to automatically control or check data without needing manual commands.

3. example
* When a new order is inserted into the orders table, a trigger automatically updates the total_sales table.
